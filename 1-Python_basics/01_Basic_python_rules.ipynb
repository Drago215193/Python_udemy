# Type Conversion
    We can convert one data type to another in some cases
        # We can convert int into string
        # We can convert number stored as string into integer 
        # We can convert float into int ex (5.11 into 5)
        # We can convert int into float ex (5 into 5.0)

        # We can't convert string with letters into int

```python
age=25
print(type(age))

# type conversion
age_str= str(age)       #used str function to change data in variable age to string and assigned new variable age_str 
print(age_str)          #print data in new variable
print(type(age_str))    #print data type of age_str
```

```python
age='25'
print(type(age))
age_int = int(age)
print(age_int)     
print(type(age_int))
```

## Dynamic Typing
Python allows the type of the variable to change as the program executes

```python
var=10 #int
print(var, type(var))

var="Hello" #str
print(var, type(var))

var=5.11 #float
print(var, type(var))
```

## Operators in Python
    

### Types of operators
    1st - Arithmetic operators
        - Addition (+)
        - Subtraction (-)
        - Multiplication (*)
        - Division (/)
        - Floor Division (//) (Removes the decimal from divided value)
        - Modulus (%) (gives the remainder)
        - Exponentials (**) (givesd power) 

```python
a=20
b=10

add= a+b          #addition
subs= a-b         #substraction
multi= a*b        #Multiplication
div= b/a          #division
floor_div= b//a   #floor division
mod= b%a          #modulus
expo= b**a       #exponentials  

print("addition is",add)
print("substraction is",subs)
print("Multiplication is",multi)
print("division is",div)
print("floor division is",floor_div)
print("modulus is",mod)
print("exponentials is",expo)
```

    2nd- Comparision Operators
        - Equal to (==)
        - Not Equal to (!=)
        - Greater than (>)
        - Greater than or equals to (>=)
        - Less than (<)
        - Less than or equals to (<=)


    3rd- Logical Operators
        - AND   (If all conditions are true then True Else False i.e if either condition is false)
        - OR    (If all conditions are False then False else if either condition is true than True )
        - NOT   Inverse of Boolean Variable (Provides opposite result of condtion, i.e. if True than output is False, If False output is True)

## Conditional statement (if, elif, else) 

### Introduction to conditional statement

    if statement
    

```python
age=20

if age>18:
    print("U are eligible to vote")
```

    Else statement
        #the else statement executes a block of code if the condition in the if statement is false

```python
age =16

if age>18:
    print("U are eligible to vote")
else:
    print("you are a minor")
```

    Elif Statement  
        # elif statement stands for "else if" 
        # elif statement allows you to check multiple statement

```python
age =22

if age>21:
    print("U are eligible to marry and vote")
elif age < 21 and age >=18:
    print("You can vote but can't marry")
else:
    print("you are a minor")
```
